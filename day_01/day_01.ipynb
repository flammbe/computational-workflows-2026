1. What is nf-core?

A service&community to share maintained reproducable nextflow pipelines which adhere to set guidelines

2. How many pipelines are there currently in nf-core?

157 as stated on the main webpage overview of the pipelines [https://nf-co.re/pipelines/].

3. Are there any non-bioinformatic pipelines in nf-core?

Yes. nf-core/datasync isn't strictly bioinformatic since it's purpose is to sync different storage locations. Those could contain bioinfo data, but it's not mandatory. [https://nf-co.re/datasync/dev/]

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A chain of different tools/commands that are executed in a specific order, to channel outputs from one process into others. This allows for more complex analysis/procedures to take place than just using one tool and chains them together in a reproducable manner. This is useful, since the subprocess of a pipeline (tools) are often reusable in other pipelines and thus don't have to be rewritten.

5. Why do you think nf-core adheres to strict guidelines?

To set standards researchers can build their work upon (as reproducability) and to allow for pipelines to be shared and used by others as well as maintained by different people.

6. What are the main features of nf-core pipelines?

- Reproducability
- Maintained
- Portability/Interoperability
- (Documentation)
- (Versioning)

#### Second run:
Duration not given, timing with package 'time' lead to total time of 10.529s.

Alo in the resumed run behind every progress step the phrase "cached" appears, indicating that the results from the last run are taken. ("[100%] 1 of 1, cached: 1 ✔")


Check out the current directory. Next to the outdir you specified, what else has changed?

In the 'work' directory there are stored hashes to the subprocesses run in the workflow. Here there were no changes. What actually was added is a new .nextflow.log.# file logging what happens in the nf run. This file states what processes were run and links to the 'work' directory. For the resumed rerun there IS a new log file added, but links to the hashes of the processes from the first run.

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!nextflow run nf-core/differentialabundance -profile test,docker -r 2.0.0 --outdir test_results -resume
```

What changed?

After deleting the work dir, the cached processes are deleted and thus are not longer cached. This results in a complete new rerun which takes the full time again. '-resume' does nothing now or more specific, checks if something is stored, fails and then just runs all processes normally. However the results of cause don't change, since nf-core pipelines are reproducable :)

### What is differential abundance analysis?

This looks at different expression levels in specific genes inbetween groups/conditions and compares them to see what genes/processes are influenced by a group/condition / to see what changes inbetween those groups.
